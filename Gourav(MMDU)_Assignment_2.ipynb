```python
import sqlite3
import pandas as pd

conn = sqlite3.connect(":memory:")
```

```python
df = pd.read_csv("Sample - Superstore.csv")
```

```python
df.to_sql(
    "superstore",
    conn,
    if_exists="replace",
    index=False,
    chunksize=100
)
```

```python
pd.read_sql("""
SELECT COUNT(*)
FROM superstore
""", conn)
```

```python
pd.read_sql(
    "PRAGMA table_info(superstore);",
    conn
)
```

```python
pd.read_sql(
    "SELECT * FROM superstore LIMIT 5;",
    conn
)
```

```python
pd.read_sql("""
SELECT *
FROM superstore
WHERE Region = 'West'
LIMIT 10;
""", conn)
```

```python
pd.read_sql("""
SELECT *
FROM superstore
WHERE Category = 'Technology'
LIMIT 10;
""", conn)
```

```python
pd.read_sql("""
SELECT *
FROM superstore
WHERE Sales > 500;
""", conn)
```

```python
pd.read_sql("""
SELECT Region,
       ROUND(SUM(Sales),2) AS Total_Sales
FROM superstore
GROUP BY Region;
""", conn)
```

```python
pd.read_sql("""
SELECT Category,
       ROUND(AVG(Sales),2) AS Avg_Sales
FROM superstore
GROUP BY Category;
""", conn)
```

```python
pd.read_sql("""
SELECT Category,
       SUM(Quantity) AS Total_Quantity
FROM superstore
GROUP BY Category;
""", conn)
```

```python
pd.read_sql("""
SELECT [Product Name],
       ROUND(SUM(Sales),2) AS Total_Sales
FROM superstore
GROUP BY [Product Name]
ORDER BY Total_Sales DESC
LIMIT 10;
""", conn)
```

```python
pd.read_sql("""
SELECT [Customer Name],
       ROUND(SUM(Sales),2) AS Total_Sales
FROM superstore
GROUP BY [Customer Name]
ORDER BY Total_Sales DESC
LIMIT 10;
""", conn)
```

```python
pd.read_sql("""
SELECT substr([Order Date],1,7) AS Month,
       ROUND(SUM(Sales),2) AS Total_Sales
FROM superstore
GROUP BY Month
ORDER BY Month;
""", conn)
```

```python
pd.read_sql("""
SELECT [Order ID],
       COUNT(*) AS Duplicate_Count
FROM superstore
GROUP BY [Order ID]
HAVING COUNT(*) > 1;
""", conn)
```

```python
pd.read_sql("""
SELECT COUNT(*) AS Total_Rows
FROM superstore;
""", conn)
```

```python
pd.read_sql("""
SELECT COUNT(*) AS Missing_Sales
FROM superstore
WHERE Sales IS NULL;
""", conn)
```
